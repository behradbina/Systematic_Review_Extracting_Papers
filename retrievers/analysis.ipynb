```python
# Read the first file and store its lines in a set
with open('pmids.txt', 'r') as f1:
    lines1 = set(line.strip() for line in f1)

# Read the second file and store its lines in a set
with open('pmcids.txt', 'r') as f2:
    lines2 = set(line[0:8].strip() for line in f2)
    
# Find the intersection (common lines) between both sets
common_lines = lines1 & lines2

# Print the common lines
for line in sorted(common_lines):
    print(line)
```

```python
with open('pmids.txt', 'r') as f1:
    lines1 = set(line.strip() for line in f1)

# Read pmcids.txt and store both full lines and first 8 characters
with open('pmcids.txt', 'r') as f2:
    pmcid_data = [(line.strip(), line[0:8].strip()) for line in f2]

# Find common first 8 characters
common_prefixes = lines1 & {prefix for (full_line, prefix) in pmcid_data}

# Get full lines from pmcids.txt that DON'T match
non_matching_lines = [full_line for (full_line, prefix) in pmcid_data 
                     if prefix not in common_prefixes]

# Save to file3.txt
with open('file3.txt', 'w') as f3:
    for line in non_matching_lines:
        f3.write(line + "/" + '\n')

print(f"Saved {len(non_matching_lines)} non-matching lines to file3.txt")
```
